```python
from google.colab import files
from PIL import Image, ImageOps
import matplotlib.pyplot as plt
import numpy as np

# Upload image
uploaded = files.upload()

filename = list(uploaded.keys())[0]

original_image = Image.open(filename).convert("RGB")

# Keep a copy for Reset
image = original_image.copy()


while True:

    print("\n==============================")
    print("IMAGE TRANSFORMATION TOOLBOX")
    print("==============================")

    print("1. Rotate")
    print("2. Resize")
    print("3. Flip")
    print("4. Shear")
    print("5. Custom Matrix")
    print("6. Reset")
    print("7. Exit")

    choice = input("Enter your choice: ")

    # Rotate
    if choice == "1":

        angle = float(input("Enter angle: "))

        image = image.rotate(angle, expand=True)

        print("Image rotated.")

    # Resize
    elif choice == "2":

        factor = float(input("Enter resize factor: "))

        width = int(image.width * factor)
        height = int(image.height * factor)

        image = image.resize((width, height))

        print("Image resized.")

    # Flip
    elif choice == "3":

        print("1. Horizontal")
        print("2. Vertical")

        flip_choice = input("Choose flip direction: ")

        if flip_choice == "1":
            image = ImageOps.mirror(image)
            print("Image flipped horizontally.")

        elif flip_choice == "2":
            image = ImageOps.flip(image)
            print("Image flipped vertically.")

    # Shear
    elif choice == "4":

        shear = float(input("Enter shear value: "))

        img_array = np.array(image)

        h, w = img_array.shape[:2]

        new_width = int(w + abs(shear) * h)

        new_image = Image.new("RGB", (new_width, h), "white")

        for y in range(h):

            shift = int(shear * y)

            for x in range(w):

                new_x = x + shift

                if 0 <= new_x < new_width:
                    new_image.putpixel(
                        (new_x, y),
                        tuple(img_array[y, x])
                    )

        image = new_image

        print("Image sheared.")

    # Custom Matrix
    elif choice == "5":

        print("Enter the 2x2 matrix")

        a = float(input("a = "))
        b = float(input("b = "))
        c = float(input("c = "))
        d = float(input("d = "))

        matrix = np.array([
            [a, b],
            [c, d]
        ])

        print("Your matrix is:")
        print(matrix)

        # PIL needs inverse transformation
        image = image.transform(
            image.size,
            Image.Transform.AFFINE,
            (
                a, b, 0,
                c, d, 0
            )
        )

        print("Custom matrix applied.")

    # Reset
    elif choice == "6":

        image = original_image.copy()

        print("Image reset to original.")

    # Exit
    elif choice == "7":

        print("Program finished.")
        break

    else:

        print("Invalid choice.")

    # Display image after operation
    plt.figure(figsize=(7, 5))
    plt.imshow(image)
    plt.axis("off")
    plt.show()
```
