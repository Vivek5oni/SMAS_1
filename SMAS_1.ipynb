```python
import numpy as np

P = np.array([
    [0.0, 0.1],
    [2.0, 1.1],
    [4.0, 1.9],
    [6.0, 3.2],
    [8.0, 3.9],
    [10.0, 5.1]
])

x = P[:, 0]
y = P[:, 1]

print(P)
print("x =", x)
print("y =", y)
print("Dimensions =", P.shape)
Dimensions = (6, 2)
```

```python
D = P[1:] - P[:-1]
norms = np.linalg.norm(D, axis=1)

print("Displacements:")
print(D)

print("Norms:")
print(norms)

print("Largest displacement =", np.max(norms))
print("Interval of largest displacement is from", np.argmax(norms),"-",(np.argmax(norms) + 1))
```

```python
import matplotlib.pyplot as plt
plt.plot(x, y, 'o-', label="Measured trajectory")

x_line = np.linspace(0, 10, 100)
y_line = 0.5 * x_line

plt.plot(x_line, y_line, label="Desired line")

plt.xlabel("x (m)")
plt.ylabel("y (m)")
plt.axis("equal")
plt.legend()
plt.grid()

plt.show()
```

Here, the drone follow the line y=0.5x with some deviation and highest deaviation occured around x=6 m according to the graph.

```python
residual = y - 0.5 * x

max_error = np.max(np.abs(residual))

print("Residuals:", residual)
print("Maximum error:", max_error)

tolerance = 0.25

if max_error <= tolerance:
    print("ACCEPT")
else:
    print("REJECT")
```

Data is same but the tolerance limit is changed.
0.2 m error is acceptable when the tolerance is 0.25 m, but 0.2 m is unacceptable when the tolerance is 0.1 m.
