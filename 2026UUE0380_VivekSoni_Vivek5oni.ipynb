```python
num = int(input("Enter an integer: "))

# Check even or odd
if num % 2 == 0:
    print("The number is even")
else:
    print("The number is odd")

# Check positive, negative or zero
if num > 0:
    print("The number is positive")
elif num < 0:
    print("The number is negative")
else:
    print("The number is zero")
```

```python
side1 = float(input("Enter first side: "))
side2 = float(input("Enter second side: "))
side3 = float(input("Enter third side: "))

# First check whether sides are positive
if side1 <= 0 or side2 <= 0 or side3 <= 0:
    print("Invalid triangle")

# Check triangle validity
elif side1 + side2 <= side3:
    print("Invalid triangle")

elif side1 + side3 <= side2:
    print("Invalid triangle")

elif side2 + side3 <= side1:
    print("Invalid triangle")

# Classify triangle
elif side1 == side2 and side2 == side3:
    print("Equilateral triangle")

elif side1 == side2 or side1 == side3 or side2 == side3:
    print("Isosceles triangle")

else:
    print("Scalene triangle")
```

```python
units = int(input("Enter electricity units consumed: "))

if units < 0:
    print("Invalid units")

elif units <= 100:
    bill = units * 2
    print("Total bill =", bill)

elif units <= 200:
    bill = (100 * 2) + ((units - 100) * 3)
    print("Total bill =", bill)

else:
    bill = (100 * 2) + (100 * 3) + ((units - 200) * 5)
    print("Total bill =", bill)
```

```python
marks = [78, 45, 91, 33, 67, 49]

passed = 0
failed = 0

for mark in marks:

    if mark >= 40:
        print(mark, "-> Pass")
        passed = passed + 1

    else:
        print(mark, "-> Fail")
        failed = failed + 1

print("Number of passed students:", passed)
print("Number of failed students:", failed)

print("Ratio of failed to passed:", failed, ":", passed)
```

```python
balance = float(input("Enter account balance: "))
withdraw = float(input("Enter amount to withdraw: "))

if withdraw <= 0:
    print("Invalid withdrawal amount")

elif withdraw > balance:
    print("Transaction cannot be completed")
    print("Insufficient balance")

else:
    balance = balance - withdraw
    print("Withdrawal successful")
    print("Remaining balance:", balance)
```

```python
amount = float(input("Enter shopping amount: "))

if amount < 0:
    print("Invalid amount")

else:

    if amount < 1000:
        discount_percent = 0

    elif amount < 5000:
        discount_percent = 10

    elif amount < 10000:
        discount_percent = 20

    else:
        discount_percent = 30

    discount = amount * discount_percent / 100
    final_amount = amount - discount

    print("Original amount:", amount)
    print("Discount:", discount)
    print("Final amount:", final_amount)
```

```python
products = {
    "laptop": {"price": 50000, "stock": 3},
    "phone": {"price": 30000, "stock": 5},
    "headphones": {"price": 2000, "stock": 0}
}

product = input("Enter product name: ")
quantity = int(input("Enter quantity: "))

if product not in products:
    print("Product is not available")

elif quantity <= 0:
    print("Invalid quantity")

elif products[product]["stock"] == 0:
    print("Product is out of stock")

elif quantity > products[product]["stock"]:
    print("Not enough stock available")

else:

    price = products[product]["price"]

    total = price * quantity

    if total > 50000:
        discount = total * 10 / 100
    else:
        discount = 0

    final_amount = total - discount

    print("Product:", product)
    print("Quantity:", quantity)
    print("Original amount:", total)
    print("Discount:", discount)
    print("Final amount:", final_amount)
```

```python
students = {
    "Aman": [85, 92, 78],
    "Riya": [95, 88, 91],
    "Karan": [72, 65, 80],
    "Simran": [45, 92, 88],
    "Raj": [90, 91, 95]
}

scholarship_count = 0

highest_average = 0
highest_scorer = ""

for name in students:

    marks = students[name]

    mark1 = marks[0]
    mark2 = marks[1]
    mark3 = marks[2]

    # Check if student failed any subject
    if mark1 < 40 or mark2 < 40 or mark3 < 40:

        category = "Failed"

    else:

        average = (mark1 + mark2 + mark3) / 3

        # Scholarship condition
        if average >= 90 and mark1 >= 85 and mark2 >= 85 and mark3 >= 85:

            category = "Scholarship"
            scholarship_count = scholarship_count + 1

        elif average >= 80:

            category = "Distinction"

        elif average >= 60:

            category = "Pass"

        else:

            category = "Needs Improvement"

        # Find highest average
        if average > highest_average:
            highest_average = average
            highest_scorer = name

    print(name, "->", category)

print()
print("Total scholarship students:", scholarship_count)
print("Highest average:", highest_average)
print("Highest scorer:", highest_scorer)
```

```python
transactions = {
    "Aman": [500, 1200, 300, 15000, 700],
    "Riya": [200, 300, 450, 600],
    "Karan": [10000, 50, 80, 12000],
    "Simran": [400, 500, 600, 700]
}

highest_total = 0
highest_customer = ""

total_suspicious = 0

for customer in transactions:

    amounts = transactions[customer]

    total = 0
    suspicious = 0

    for amount in amounts:

        total = total + amount

        if amount > 10000:
            suspicious = suspicious + 1

    # Check risk
    if suspicious >= 2:
        risk = "High Risk"

    elif suspicious == 1 and total > 20000:
        risk = "High Risk"

    else:
        risk = "Low Risk"

    print(customer, "->", risk)
    print("Total amount spent:", total)
    print("Suspicious transactions:", suspicious)
    print()

    # Find highest spending customer
    if total > highest_total:
        highest_total = total
        highest_customer = customer

    # Count all suspicious transactions
    total_suspicious = total_suspicious + suspicious


print("Highest spending customer:", highest_customer)
print("Highest total spending:", highest_total)
print("Total suspicious transactions:", total_suspicious)
```
